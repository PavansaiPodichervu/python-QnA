# Batch 1 — Lists (Q1–Q10)

## Q1. Remove duplicates while preserving order

**Scenario:** An API returns duplicate stock symbols. Preserve first occurrence order.

```python
stocks = ["TCS", "INFY", "TCS", "RELIANCE", "INFY", "HDFC"]
```
**Answer**
```python
def unique_preserve_order(items):
    seen = set()
    result = []
    for item in items:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result

unique_preserve_order(stocks)
# ['TCS', 'INFY', 'RELIANCE', 'HDFC']
```
**Explanation:** A set provides average O(1) membership checks; the result list preserves order.  
**Complexity:** O(n) time, O(n) extra space. Requires hashable elements.

## Q2. Find the second-largest distinct number

```python
values = [10, 5, 10, 8, 8, 3]
```
**Answer**
```python
def second_largest(values):
    distinct = set(values)
    if len(distinct) < 2:
        return None
    return sorted(distinct, reverse=True)[1]

second_largest(values)  # 8
```
**Explanation:** Deduplicate before ranking; otherwise repeated maximum values can incorrectly become the “second” result.  
**Complexity:** O(n log n) time. An O(n) one-pass solution is possible when optimizing further.

## Q3. Rotate a list right by k positions

```python
items = [1, 2, 3, 4, 5]
k = 2
```
**Answer**
```python
def rotate_right(items, k):
    if not items:
        return []
    k %= len(items)
    return items[-k:] + items[:-k] if k else items[:]

rotate_right(items, 2)  # [4, 5, 1, 2, 3]
```
**Explanation:** Modulo handles rotations larger than the list length; the k==0 branch avoids the `[-0:]` slicing trap.  
**Complexity:** O(n) time and O(n) space.

## Q4. Flatten a one-level nested list

```python
data = [[1, 2], [3], [], [4, 5]]
```
**Answer**
```python
flat = [item for group in data for item in group]
# [1, 2, 3, 4, 5]
```
**Explanation:** The outer loop iterates groups; the inner loop emits each item. This is one-level flattening, not recursive flattening.  
**Complexity:** O(n) time and O(n) output space.

## Q5. Split a list into chunks

```python
items = [1, 2, 3, 4, 5, 6, 7]
size = 3
```
**Answer**
```python
def chunks(items, size):
    if size <= 0:
        raise ValueError("size must be positive")
    return [items[i:i + size] for i in range(0, len(items), size)]

chunks(items, 3)  # [[1, 2, 3], [4, 5, 6], [7]]
```
**Explanation:** Slice boundaries advance by the chunk size. Validate size to avoid an invalid range step.  
**Complexity:** O(n) time and O(n) space.

## Q6. Find common elements while preserving the first list's order

```python
a = [4, 2, 4, 1, 3]
b = [3, 4, 4]
```
**Answer**
```python
b_set = set(b)
common = list(dict.fromkeys(x for x in a if x in b_set))
# [4, 3]
```
**Explanation:** Set membership is efficient; `dict.fromkeys` removes repeated matches while preserving encounter order.  
**Complexity:** Average O(n + m) time, O(m + output) space.

## Q7. Remove falsy values—but preserve zero when required

## Q8. Sort records by score descending, then name ascending

```python
users = [{"name": "Zed", "score": 90},
         {"name": "Amy", "score": 90},
         {"name": "Bob", "score": 80}]
```
**Answer**
```python
users_sorted = sorted(users, key=lambda x: (-x["score"], x["name"]))
# Amy, Zed, Bob
```

**Complexity:** O(n log n) time.

## Q10. Avoid quadratic list concatenation

**Problem:** Why is this inefficient?
```python
result = []
for batch in batches:
    result = result + batch
```
**Answer**
```python
result = []
for batch in batches:
    result.extend(batch)
# or: result = [item for batch in batches for item in batch]
```
**Explanation:** `result + batch` allocates a new list and copies the accumulated prefix on every iteration. `extend` mutates the existing list.  
**Complexity:** Repeated concatenation can become O(n²); extend across all items is O(total items).

# Batch 2 — Dictionaries (Q11–Q20)

## Q11. Group records by symbol

```python
records = [{"symbol":"TCS","price":10}, {"symbol":"INFY","price":20},
           {"symbol":"TCS","price":11}]
```
**Answer**
```python
from collections import defaultdict
grouped = defaultdict(list)
for row in records:
    grouped[row["symbol"]].append(row)
grouped = dict(grouped)
```
**Explanation:** `defaultdict(list)` initializes a list on first access.  
**Complexity:** O(n) time and O(n) space.

## Q15. Invert a mapping when values are duplicated

```python
mapping = {"a": 1, "b": 2, "c": 1}
```
**Answer**
```python
from collections import defaultdict
inverted = defaultdict(list)
for key, value in mapping.items():
    inverted[value].append(key)
# {1: ['a', 'c'], 2: ['b']}
```
**Explanation:** A simple `{v: k for k, v in mapping.items()}` loses keys when values repeat.

## Q16. Keep the latest record per key

```python
rows = [{"id":"A","ts":2}, {"id":"B","ts":1}, {"id":"A","ts":5}]
```
**Answer**
```python
latest = {}
for row in rows:
    key = row["id"]
    if key not in latest or row["ts"] > latest[key]["ts"]:
        latest[key] = row
# A -> ts 5, B -> ts 1
```
**Explanation:** One pass compares each timestamp against the current winner.  
**Complexity:** O(n) time; O(number of unique IDs) space.

## Q17. Dictionary lookup with missing keys

**Question:** Compare `d[key]`, `d.get(key)`, and `key in d`.
- `d[key]`: returns value or raises `KeyError`.
- `d.get(key, default)`: returns default if absent; does not insert.
- `key in d`: checks key presence, not value truthiness.
**Important:** A stored value of `None` is different from a missing key.

## Q18. Use setdefault without an expensive default

**Risky**
```python
cache.setdefault(key, expensive_computation())
```
`expensive_computation()` executes before `setdefault` is called, even if key exists.
**Better**
```python
if key not in cache:
    cache[key] = expensive_computation()
value = cache[key]
```
**Explanation:** Avoid unnecessary work. In concurrent code, this check-then-set may need synchronization.

## Q19. Build an index for repeated lookups

**Scenario:** Find records by ID many times.

**Explanation:** Build once in O(n), then average O(1) lookup. Duplicate IDs overwrite earlier records; choose a policy if duplicates are possible.

```python
seen = set()
unique = []
for row in rows:
    key = (row["symbol"], row["exchange"])
    if key not in seen:
        seen.add(key)
        unique.append(row)
```

# Batch 4 — Advanced Patterns & Performance (Q31–Q40)

## Q31. Explain list vs tuple vs set vs dict

- **list:** ordered, mutable sequence; duplicates allowed; index access O(1).
- **tuple:** ordered, immutable sequence; hashable only if all elements are hashable.
- **set:** unique elements; average O(1) membership; no positional indexing.
- **dict:** key-value mapping; insertion ordered; average O(1) key lookup.
**Selection:** Choose based on semantics first, then performance.

## Q32. Why can a mutable default argument be dangerous?

**Bad**
```python
def add_item(item, items=[]):
    items.append(item)
    return items
```
The same list is reused across calls.
**Correct**
```python
def add_item(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items
```
**Explanation:** Default arguments are evaluated once when the function is defined, not on each call.

## Q33. What is the difference between `is` and `==`?

- `==` checks value equality.
- `is` checks object identity.
```python
a = [1, 2]; b = [1, 2]
a == b  # True
a is b  # False
```
Use `is None` for None checks. Do not rely on identity for numeric/string values due to implementation optimizations.

## Q35. Why are sets useful for membership-heavy workflows?

**Bad**

**Better**

## Q36. Explain Timsort and why it is useful

Python's sort uses **Timsort**, a stable hybrid sorting algorithm that exploits existing ordered runs in data.
- Worst-case time: O(n log n).
- Best-case on highly ordered data: O(n).
- Stable: equal keys preserve input order.
**Practical implication:** Sorting partially ordered event or timestamp data can be faster than the worst-case bound suggests.

## Q37. How do you avoid repeated computation?

Use a cache when inputs are repeatable and results are safe to reuse:
```python
from functools import lru_cache
@lru_cache(maxsize=256)
def normalize_symbol(symbol):
    return symbol.strip().upper()
```
**Trade-offs:** Cache memory, invalidation, stale results, hashable arguments, and concurrency behavior. Cache only deterministic functions or define freshness rules.

## Q39. When should you use deque, heap, or set?

- **deque:** FIFO/LIFO operations at both ends.
- **heapq:** repeatedly retrieve smallest/largest priority item; scheduling/top-k.
- **set:** uniqueness and membership.
- **dict:** keyed lookup and mapping.
**Interview expectation:** Explain why the data structure matches operation patterns, not just its definition.

## Q49. Design a bounded in-memory cache

```python
from collections import OrderedDict
class LRU:
    def __init__(self, capacity):
        self.capacity = capacity
        self.data = OrderedDict()
    def get(self, key):
        if key not in self.data:
            return None
        self.data.move_to_end(key)
        return self.data[key]
    def put(self, key, value):
        self.data[key] = value
        self.data.move_to_end(key)
        if len(self.data) > self.capacity:
            self.data.popitem(last=False)
```
**Explanation:** LRU means least recently used. This sketch supports average O(1) operations; production code should clarify missing-value semantics, capacity validation, thread safety, and TTL/invalidation.
